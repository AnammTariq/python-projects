```python
#Major code :2
def select_user(Current_balance):
    print("To Check Balance Enter choice 1")
    print("To Deposite Money Enter choice 2")
    print("To Withdraw Money Enter choice 3")
    print("To Exit Enter choice 4")
    while 1:
     choice=int(input("enter your choice: "))
     if choice==1:
        print(f"Current Balance is: RS{Current_balance}")
        print(Current_balance)
     elif choice==2:
        amount=int(input("enter the amount you want to deposit"))
        Current_balance=Current_balance+amount
        print(f"{amount} is deposite succesfully.the new balance is {Current_balance}")
        #print(Current_balance)
     elif choice ==3:
        amount=int(input("enter the amount you want to withdraw"))
        if amount>Current_balance:
            print(f"insufficent balance!")
        else:
            Current_balance=Current_balance-amount
            print(f"{amount} is withdraw succesfully. the new balance is {Current_balance}")
     elif choice==4:
        print("Thank you for using the ATM")
        break   
    else:
        print(f"invalid menu choice. select option betwwen 1 to 4")
```

```python
def main(): 
    Current_balance=10000
    Current_balance=select_user(Current_balance)
```
